```python
#========================================
#DAY 2 PROJECT:STUDENT GRADE CALCULATOR
#========================================

# STEP 1: Get student  information

name = input("Enter student name ")

python_marks = int(input("Enter python Marks"))
SQL_marks = int(input("Enter SQL marks"))
statistics_marks = int(input(" Enter statistics marks"))

# STEP 2: Validate marks

if not (0 <= python_marks < 100 and 0 <= SQL_marks <100 and 0<=  statistics_marks <100 ):
        print("Invalid marks! Enter values between 0 and 100")
else :
  # STEP 3: Calculate total and average
  total = python_marks + SQL_marks + statistics_marks
  average = total / 3

  # STEP 4 : Determine result
  if python_marks < 35 or SQL_marks < 35 or statistics_marks<35:
    result = "FAIL"
    grade = "F"

  else:
    result = "Pass"

    # STEP 5: assign grade
    if average >= 90:
      grade = "A+"
    elif average >=75:
      grade = "A"
    elif average >= 50:
      grade ="B"
    else:
      grade = "C"

    #STEP 6: Display report

    print("\n===========STUDENT REPORT=============")

    print("student Name:",name)
    print("Python marks", python_marks)
    print("SQL Marks", SQL_marks)
    print("statistics Marks:", statistics_marks)

    print("-------------------------")

    print("TOTAL Marks", total)
    print("Average Marks:", round(average,2))
    print("Result :", result)
    print("Grade:",grade)

    print("========================================")
```
