```python
# DAY 3 project: STUDENT MARKS ANALYZER

marks = [85,45,92,30,76,65,55,95]

total = 0
highest = marks[0]
lowest = marks[0]
passed = 0
failed = 0

for mark in marks:

    #calculate total
    total = total + mark

    #find lowest mark
    if mark < lowest:
        lowest= mark

    # Find highest mark
    if mark < highest:
        highest= mark

    # Count pass  or fail 
    if mark >= 35:
        passed +=1
    else:
        failed += 1

average = total/len(marks)

print("======STUDENT  MARKS REPORT =====")
print("total Marks:", total)
print("average Marks:", round(average,2))
print("Highest Marks:", highest)
print("Lowest Marks:", lowest)
print("student Pass", passed)
print("Student Fail", failed)
print("========================================")
```
