```python
 # DAY 4 PROJECT : STUDENT PERFORMANCE ANALYZER

# Student marks
marks = [85,45, 92,30,76,65,55,95]
# FUNCTION 1: calculate total
def calculate_total(marks):
    return sum(marks)

# FUNCTION 2: Calculate average 
def calculate_average(marks):
    return calculate_total(marks)/ len(marks)

# FUNCTION 3: Find highest mark 
def find_highest(marks):
    highest = marks[0]

    for mark in marks:
        if mark > highest:
            highest = mark
       
    return highest

#FUNCTION 4: Find lowest mark
def find_lowest(marks):
    lowest = marks[0]

    for mark in marks:
        if mark < lowest:
            lowest = mark

    return lowest

# FUNCTION 5:Count passed students
def count_passed(marks):
    count =0
    for mark in marks:
        if mark >= 35:
            count += 1
    
    return count 

# FUNCTION 6: Count failed student
def count_failed(marks):
    count = 0 
    for mark in marks:
        if mark <= 35:
            count += 1

    return count

#Display report
if len(marks)>0:
    print("===== STUDENT PERFORMANCE REPORT =====")

    print("Total Marks:", calculate_total(marks))
    print("Average Marks:", round(calculate_average(marks), 2))
    print("Highest Mark:", find_highest(marks))
    print("Lowest Mark:", find_lowest(marks))
    print("Passed Students:", count_passed(marks))
    print("Failed Students:", count_failed(marks))

    print("======================================")
else:
    print("No student marks available.")  
```
